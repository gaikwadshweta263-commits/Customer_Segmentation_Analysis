## Step 1:- Data Loading

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

customers = pd.read_csv("/content/customers.csv")
transactions = pd.read_csv("/content/transactions.csv")
```

## Step 2: Understanding the Datasets


```python
print("Customers Information:")
customers.info()

print("\nTransactions Information:")
transactions.info()
```

```python
print("Customers Dataset")
display(customers.head())

print("Transactions Dataset")
display(transactions.head())
```

## Objective :-

- Understand Data & its type.
- Identify numerical & categorical columns.
- Dectect missing values.
- understand stat's representation.

```python
print("Customer Statistics:")
display(customers.describe(include="all").T)

print("Transaction Statistics:")
display(transactions.describe(include="all").T)
```

# Data Cleaning

- Removing Irrelevant Columns

```python
customers.drop(columns=["Unnamed: 0"], errors="ignore", inplace=True)
transactions.drop(columns=["Unnamed: 0"], errors="ignore", inplace=True)

print("Unnecessary columns removed.")
```

- Missing Data Detection

```python
print("Missing Values in Customers:")
print(customers.isnull().sum())

print("\nMissing Values in Transactions:")
print(transactions.isnull().sum())
```

- Text Data Cleaning

```python
customers["Gender"] = customers["Gender"].str.strip().str.title()
customers["City"] = customers["City"].str.strip().str.title()
customers["State"] = customers["State"].str.strip().str.title()

print(customers.head())
```

- Check Duplicates

```python
print("Duplicate customer rows:", customers.duplicated().sum())
print("Duplicate transaction rows:", transactions.duplicated().sum())

customers.drop_duplicates(inplace=True)
transactions.drop_duplicates(inplace=True)
```

- Final Data Quality Check

```python
print("Customers Missing Values:")
print(customers.isnull().sum())

print("\nTransactions Missing Values:")
print(transactions.isnull().sum())
```

## Step 3: Exploratory Data Analysis (EDA)

### Q1. What is the distribution of customer ages?

```python
plt.figure()
plt.hist(customers["Age"].dropna(), bins=20)
plt.title("Distribution of Customer Ages")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

### Analysis:
- The histogram shows the distribution of customers across different age groups.

### Conclusion:
- Understanding customer ages helps businesses create suitable marketing strategies for different age groups.

### Q2. Which gender has the highest number of customers?


```python
plt.figure()
sns.countplot(data=customers, x="Gender")
plt.title("Customer Distribution by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Customers")
plt.show()
```

### Analysis:
- The count plot compares the number of customers in each gender category.

### Conclusion:
- Gender distribution helps businesses understand their customer demographics.

### Q3. Which cities have the most customers?


```python
top_cities = customers["City"].value_counts().head(10)
plt.figure(figsize=(10, 5))
sns.barplot(x=top_cities.index, y=top_cities.values)
plt.title("Top 10 Cities by Customer Count")
plt.xlabel("City")
plt.ylabel("Number of Customers")
plt.xticks(rotation=45)
plt.show()
```

### Analysis:
- The bar chart displays the top 10 cities with the highest number of customers.

### Conclusion:
- Businesses can focus their marketing campaigns on cities with a large customer base

### Q4. What is the distribution of transaction values?



```python
plt.figure()
plt.hist(transactions["TotalValue"].dropna(), bins=30)
plt.title("Distribution of Transaction Values")
plt.xlabel("Total Transaction Value")
plt.ylabel("Frequency")
plt.show()
```

### Analysis:
- The histogram shows the frequency of transactions across different purchase amounts.

### Conclusion:
- Understanding transaction values helps businesses identify common spending ranges and purchasing patterns.

### Q5. How does quantity purchased affect transaction value?




```python
plt.figure()
plt.scatter(transactions["Quantity"],transactions["TotalValue"])
plt.title("Quantity vs Transaction Value")
plt.xlabel("Quantity Purchased")
plt.ylabel("Total Transaction Value")

plt.show()
```

### Analysis:
- The scatter plot shows the relationship between quantity purchased and total transaction value.

### Conclusion:
- The relationship helps businesses understand bulk purchasing behavior and evaluate quantity-based offers.

### Q6. Which products generate the highest transaction value?


```python
top_products = (transactions.groupby("ProductID")["TotalValue"].sum().sort_values(ascending=False).head(10))
plt.figure(figsize=(10, 5))
sns.barplot(x=top_products.index, y=top_products.values)
plt.title("Top 10 Products by Transaction Value")
plt.xlabel("Product ID")
plt.ylabel("Total Transaction Value")
plt.xticks(rotation=45)
plt.show()
```

### Analysis:
- The bar chart identifies the top 10 products based on total transaction value.

### Conclusion:
- Businesses can identify products that contribute significantly to recorded sales value.

### Q7. Which months have the highest transaction value?


```python
transactions["TransactionDate"] = pd.to_datetime(transactions["TransactionDate"],dayfirst=True,errors="coerce")
monthly_sales = (transactions.dropna(subset=["TransactionDate"]).groupby(transactions["TransactionDate"].dt.to_period("M"))["TotalValue"].sum())
plt.figure(figsize=(12, 5))
plt.plot(monthly_sales.index.astype(str), monthly_sales.values, marker="o")
plt.title("Monthly Transaction Value Trend")
plt.xlabel("Month")
plt.ylabel("Total Transaction Value")
plt.xticks(rotation=45)
plt.show()
```

### Q8. What is the relationship between product price and quantity?


```python
plt.figure()
sns.scatterplot(data=transactions,x="Price",y="Quantity")
plt.title("Product Price vs Quantity")
plt.xlabel("Price")
plt.ylabel("Quantity Purchased")
plt.show()
```

### Analysis:
- The scatter plot compares product prices with the quantity purchased in each transaction.

### Conclusion:
- This analysis helps businesses explore whether purchasing quantities differ across price levels.

### Q9. Which states have the highest number of customers?


```python
top_states = customers["State"].value_counts().head(10)
plt.figure(figsize=(10, 5))
sns.barplot(x=top_states.values, y=top_states.index)
plt.title("Top 10 States by Customer Count")
plt.xlabel("Number of Customers")
plt.ylabel("State")
plt.show()
```

### Analysis:
- The bar chart compares customer counts across the top 10 states.

### Conclusion:
- Geographical customer analysis helps businesses identify important markets and expansion opportunities.

### Q10. What is the correlation between quantity, price, and transaction value?


```python
corr = transactions[["Quantity", "Price", "TotalValue"]].corr()
plt.figure(figsize=(7, 5))
sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Transaction Correlation Heatmap")
plt.show()
```

### Analysis:
- The correlation heatmap displays the strength and direction of relationships among numerical transaction columns.

### Conclusion:
- Correlation analysis helps identify relationships between purchasing quantity, product price, and total transaction value.
